```python
# Python program to explain cv2.rotate() method

# importing cv2
import cv2
import numpy

# path
path = 'dalaam_square.png'

# Reading an image in default mode
src = cv2.imread(path,0)

type(src)

# Window name in which image is displayed
window_name = 'Image'

# Using cv2.rotate() method
# Using cv2.ROTATE_90_CLOCKWISE rotate
# by 90 degrees clockwise
resize = cv2.resize(src, (2048, 2048))


#b,g,r = cv2.split(resize)

# Filename
# filename = 'original.bmp'

cv2.imwrite('./results/original.bmp', resize)


# Displaying the image
cv2.imshow(window_name, resize)
cv2.waitKey(0)
```

```python
# Write resized image to raw txt for OpenCL program to run
array = numpy.array(resize)
h, w = array.shape[:2]
top = "(" + str(w) + ", " + str(h) + ")"
numpy.savetxt("./improc/input.raw",array, fmt='%d', delimiter=' ', header=top)
```

```python
# Read output as image !
result_blur = numpy.loadtxt('./improc/outputs/blur.txt', skiprows=1, dtype=numpy.uint8)
result_sharpen = numpy.loadtxt('./improc/outputs/sharpen.txt', skiprows=1, dtype=int)
result_h_edge = numpy.loadtxt('./improc/outputs/h_edge.txt', skiprows=1, dtype=int)
result_v_edge = numpy.loadtxt('./improc/outputs/v_edge.txt', skiprows=1, dtype=int)

cv2.imwrite('./results/blur.bmp',result_blur)
cv2.imwrite('./results/sharpen.bmp',result_sharpen)
cv2.imwrite('./results/h_edge.bmp',result_h_edge)
cv2.imwrite('./results/v_edge.bmp',result_v_edge)

# cv2.imshow(window_name, result)
# cv2.waitKey(0)
```
